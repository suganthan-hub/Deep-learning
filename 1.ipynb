```python
import numpy as np
import matplotlib.pyplot as plt

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Addition:\n", A + B)
print("Multiplication:\n", A @ B)
print("Transpose:\n", A.T)

plt.imshow(A, cmap="viridis")
plt.colorbar()
plt.title("Matrix A")
plt.show()
```
